```python
import pandas as pd
import re

df = pd.read_csv(
    "collections/SMSSpamCollection",
    sep="\t",
    header=None,
    names=["label", "message"],
)

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
    return text

df.head(100)

df["message"] = df["message"].apply(clean_text)
df["message"]
```
