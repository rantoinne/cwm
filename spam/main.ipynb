```python
import re
import pandas as pd
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS, CountVectorizer, TfidfVectorizer

df = pd.read_csv(
    "collections/SMSSpamCollection",
    sep="\t",
    header=None,
    names=["label", "message"],
)

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
    text = " ".join(w for w in text.split() if w not in ENGLISH_STOP_WORDS)
    return text

df["message"] = df["message"].apply(clean_text)
df["label"] = df["label"].map({ "ham": 0, "spam": 1 })
df["message"]
```

```python
# vectorizer = CountVectorizer(token_pattern=r"(?u)\b[a-zA-Z]{2,}\b")
vectorizer = CountVectorizer()

x_sparse = vectorizer.fit_transform(df["message"])

feature_names = vectorizer.get_feature_names_out()

df_counts = pd.DataFrame(x_sparse.toarray(), columns=feature_names)

df_counts.shape
# df_counts
```

```python
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(df["message"])

tfidf_counts = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf.get_feature_names_out())

# tfidf_counts.shape

# tfidf_counts.loc[0, tfidf_counts.loc[0] > 0]

# tfidf_counts["free"].sum()

tfidf_counts
```
